# Business Analysis Challenges


# Challenge 1
Identify the city generating the highest revenue.


```python
cities= df.groupby("City")["Amount_Spent"].sum()
highest_revenue=cities.idxmax()
print(highest_revenue)
```

# Challenge 2
Identify the category generating the highest revenue.


```python
Categeries= df.groupby("Category")["Amount_Spent"].sum()
highest_Category=Categeries.idxmax()
print(highest_Category)
```

# Challenge 3
Determine which gender spends more on average.


```python
genders=df.groupby("Gender")["Amount_Spent"].mean()
print(genders)
```

# Challenge 4
Identify the category with the highest average customer rating.


```python
cetegories=df.groupby("Category")["Rating"].mean()
print(cetegories.idxmax())
```

# Challenge 5
Identify the category receiving the highest number of orders


```python
cetogories=df.groupby("Category")["Orders"].sum()
highest_orders=cetogories.idxmax()
print(highest_orders)
```

 # Challenge 6
Identify which gender contributes the highest total revenue.


```python
genders=df.groupby("Gender")["Amount_Spent"].sum()
print("Highest  revenue  gender:",genders.idxmax())
```
